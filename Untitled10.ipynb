```python
import re
import tweepy
from tweepy import OAuthHandler
from textblob import TextBlob

class TwitterClient(object):
    '''
    Generic Twitter Class for sentiment analysis.
    '''
    def __init__(self):
        '''
        Class constructor or initialization method.
        '''
        # keys and tokens from the Twitter Dev Console
        consumer_key = 'Pmo3RtXq6cZAbaVEp3TzxoZnf'
        consumer_secret = '4Bb0Fs9azQqSR1uiKI81sc1ktnhhoCdio8ZlPMT0WqmyPDm82y'
        access_token = '1912905453027360768-gsxSVs913jDzTh0ASed7GCHQhWABsR'
        access_token_secret = '05rGX5lwy7QRsigbi9ZA5vhvQI3xmlIDsstsM9q1LHaUO'

        try:
            self.auth = OAuthHandler(consumer_key, consumer_secret)
            self.auth.set_access_token(access_token, access_token_secret)
            self.api = tweepy.API(self.auth)
        except Exception as e:
            print("Error: Authentication Failed -", e)

    def clean_tweet(self, tweet):
        '''
        Clean tweet text by removing links, special characters using regex.
        '''
        return ' '.join(re.sub(r"(@[A-Za-z0-9_]+)|([^0-9A-Za-z \t])|(\w+:\/\/\S+)", " ", tweet).split())

    def get_tweet_sentiment(self, tweet):
        '''
        Classify sentiment of a tweet using TextBlob.
        '''
        analysis = TextBlob(self.clean_tweet(tweet))
        if analysis.sentiment.polarity > 0:
            return 'positive'
        elif analysis.sentiment.polarity == 0:
            return 'neutral'
        else:
            return 'negative'

    def get_tweets(self, query, count=10):
        '''
        Fetch tweets and parse them.
        '''
        tweets = []

        try:
            # Using updated method: search_tweets (not deprecated 'search')
            fetched_tweets = self.api.search_tweets(q=query, count=count)

            for tweet in fetched_tweets:
                parsed_tweet = {
                    'text': tweet.text,
                    'sentiment': self.get_tweet_sentiment(tweet.text)
                }

                if tweet.retweet_count > 0:
                    if parsed_tweet not in tweets:
                        tweets.append(parsed_tweet)
                else:
                    tweets.append(parsed_tweet)

            return tweets

        except tweepy.errors.TweepyException as e:
            print("Error:", str(e))


def main():
    # create object of TwitterClient
    api = TwitterClient()

    # fetch tweets
    tweets = api.get_tweets(query='Donald Trump', count=200)

    if not tweets:
        print("No tweets fetched.")
        return

    # filter by sentiment
    ptweets = [tweet for tweet in tweets if tweet['sentiment'] == 'positive']
    ntweets = [tweet for tweet in tweets if tweet['sentiment'] == 'negative']

    print("Positive tweets percentage: {:.2f}%".format(100 * len(ptweets) / len(tweets)))
    print("Negative tweets percentage: {:.2f}%".format(100 * len(ntweets) / len(tweets)))
    print("Neutral tweets percentage: {:.2f}%".format(100 * (len(tweets) - len(ptweets) - len(ntweets)) / len(tweets)))

    print("\n\nPositive tweets:")
    for tweet in ptweets[:10]:
        print(tweet['text'])

    print("\n\nNegative tweets:")
    for tweet in ntweets[:10]:
        print(tweet['text'])


if __name__ == "__main__":
    main()
```

```python
import requests
from textblob import TextBlob
import re

# Replace with your NewsAPI key
API_KEY = '3ac96ee4947241968d38097a1cf3f6aa'

def clean_text(text):
    return ' '.join(re.sub(r"(@[A-Za-z0-9_]+)|([^0-9A-Za-z \t])|(\w+:\/\/\S+)", " ", text).split())

def get_sentiment(text):
    analysis = TextBlob(clean_text(text))
    if analysis.sentiment.polarity > 0:
        return 'positive'
    elif analysis.sentiment.polarity == 0:
        return 'neutral'
    else:
        return 'negative'

def fetch_news(query, api_key=API_KEY):
    url = f'https://newsapi.org/v2/everything?q={query}&language=en&pageSize=20&apiKey={api_key}'
    response = requests.get(url)
    data = response.json()

    articles = data.get('articles', [])
    results = []

    for article in articles:
        title = article['title']
        sentiment = get_sentiment(title)
        results.append({'title': title, 'sentiment': sentiment})

    return results

# Example usage
query = "India"
articles = fetch_news(query)

positive = [a for a in articles if a['sentiment'] == 'positive']
negative = [a for a in articles if a['sentiment'] == 'negative']
neutral = [a for a in articles if a['sentiment'] == 'neutral']

print(f"Positive: {len(positive)}")
print(f"Negative: {len(negative)}")
print(f"Neutral: {len(neutral)}")

print("\nSample Positive Articles:")
for a in positive[:3]:
    print("-", a['title'])

print("\nSample Negative Articles:")
for a in negative[:3]:
    print("-", a['title'])
```

```python
CATEGORY_KEYWORDS = {
    'Politics': ['election', 'government', 'parliament', 'minister', 'policy', 'modi', 'biden', 'president'],
    'Technology': ['tech', 'apple', 'microsoft', 'google', 'software', 'hardware', 'ai', 'robot', 'iphone'],
    'Sports': ['cricket', 'football', 'soccer', 'tennis', 'olympics', 'match', 'team', 'score'],
    'Business': ['market', 'stock', 'economy', 'trade', 'startup', 'profit', 'business'],
    'Health': ['covid', 'health', 'vaccine', 'virus', 'hospital', 'medicine', 'disease'],
    'Entertainment': ['movie', 'film', 'celebrity', 'netflix', 'bollywood', 'series', 'music']
}
```

```python
def classify_category(text):
    text = text.lower()
    for category, keywords in CATEGORY_KEYWORDS.items():
        if any(word in text for word in keywords):
            return category
    return 'Other'
```

```python
results = []  # ← Initialize the list before using it
for article in articles:
    title = article['title']
    sentiment = get_sentiment(title)
    category = classify_category(title)
    results.append({'title': title, 'sentiment': sentiment, 'category': category})
```

```python
from collections import Counter
categories = [a['category'] for a in results]
category_counts = Counter(categories)

print("\nCategory Distribution:")
for cat, count in category_counts.items():
    print(f"{cat}: {count}")
```
